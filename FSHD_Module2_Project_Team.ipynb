## Disease Background:

* Prevalence & incidence: The prevalence rate of FSHD is 1 case per 8,000. The incidence, based on population studies in the Netherlands, is 0.3/1000,0000 persons-years, or 52 diagnosed people in the Netherlands. However, more cases remain misdiagnosed or undiagnosed.  (Deenen et al., 2014; Facioscapulohumeral Muscular Dystrophy (FSHD) - Diseases, 2015)
* Symptoms: Facial weakness is often the earliest sign of FSHD, making it hard to use a straw or smile. Other hallmark symptoms include shoulder weakness, causing scapular winging, and lower leg weakness, causing a foot drop which increases the risk of tripping and difficulty walking. Additionally, chronic pain is common in FSHD patients due to strain related to posture and weakness. (Signs and Symptoms of Facioscapulohumeral Muscular Dystrophy (FSHD) - Diseases, 2015)
* Diagnosis: Diagnosis of FSHD is usually suspected based on a pattern of muscle weakness, typically in the face shoulder blades and upper arms, often with asymmetric involvement. The most reliable way to confirm a diagnosis is through genetic testing. A contraction of the D4Z4 region on chromosome 4 identifies this disease. In some cases where a diagnosis is unclear, blood tests that measure creatine kinase, EMGs, or muscle biopsies can also be used. (Diagnosis - Facioscapulohumeral Muscular Dystrophy (FSHD) - Diseases, 2015)

* Standard of care treatments (& reimbursement): Patients suffering from FSHD often begin occupational and/or physical therapy in order to adapt performance daily tasks manage pain and other symptoms. Occupational therapists often observe the patient in their home environment in order to suggest ways of improving efficiency and comfort in doing tasks or the use of assistive technology such as wheelchairs. Physical therapists work with patients to perform exercises to rehabilitate their muscles. Physical therapy is often prescribed by primary care doctors when a patient's FSHD symptoms progress more quickly than usual, in order to adapt to their worsened muscle strength. Further, scapular fixation surgery is a common procedure that attaches the scapula to the ribcage in order to stabilize the shoulder. The surgery is performed on patients that no longer can raise their arm above their shoulder, which greatly hinders daily activities. (FSHD Clinical Care - FSHD Society, 2019)
* Disease progression & prognosis: FSHD is a slow progressing disease that gradually deteriorates the muscle fibers in the body, causing muscle weakness. Per the name of the disease, the face, shoulder, and upper arm muscle regions often degenerate the most, although all muscles of the body can be affected. Disease progression rate and severity varies by individual. In rare cases, patients with FSHD develop respitory, hearing, or vissoin issues, especially when the disease is early-onset and begins in childhood. (Facioscapulohumeral Muscular Dystrophy (FSHD) - Diseases, 2015)

* Biological mechanisms (anatomy, organ physiology, cell & molecular physiology): FSHD is an epigenetic condition due to the 4q35 region in chromosome 4. This region is a modulator of DUX4 activity. Although the DUX4 gene is important for embryological development, in healthy adults the gene shuts down due to D4Z4 array, which prevents DUX4 from being transcribed any further. A healthy D4Z4 array typically has more than 10 repeats, causing it to be longer and more compact. However, in FSHD patients, the D4Z4 array has less than 10 repeats, causing it to be less compact. In FSHD1, the D4Z4 chromosomal tandem repeats are deleted, causing the continued transcription of the DUX4 gene and therefor the secretion of the DUX4 protein. In FSHD2, the D4Z4 chromatin has repeats but is not contracted due to a dysfunctional SMCHD1 (structural maintenance of chromosomes flexible hinge domain containing 1) protein, also causing the transcription and secretion of DUX4. The over-expression of DUX4 leads to progressive muscle degeneration due to muscle cell apoptosis or impairment of muscle differentiation. This eventually leads to muscle fiber loss and fatty infiltration. The expression of DUX4 is not constant in all muscles in the body, causing some tissues to deteriorate more quickly than others. (Blemker, 2026; Genetic Cause - FSHD Society, 2025)

* Clinical Trials/next-gen therapies: Research on molecular, biological, and gene therapies are being conducted to slow or stop FSHD progression. The main goal of the drugs are to restore immune balance, support muscles, or block DUX4 expression. Although a majority if the drugs are still in the research phase, three are in clinical phase 1 (Restem-L / ULSC, Apitegromab (SRK-015), and Clenbuterol), three are in phase 2 (Satralizumab, ARO-DUX4 / SRP-1001, and EPI-321), and one is in phase 3 (Del-brax / AOC 1020). While most therapies aim to block the DUX4 gene, research is also being conducted on supplement development and physical therapy practices. (Drug Development Pipeline - FSHD Society, 2021)

Why interpolation is an estimate rather than a new measurement
Interpolation is an estimate because it uses existing data to predict values at locations where data is missing.
- The assumptions made about how the measurement changes between sampled locations
It assumes that the values follow the general pattern of the rest of the data set.
- Whether the estimate is within the range of sampled data or represents extrapolation
The estimate is considered an interpolation if the value falls within the range of given data. It is considered extrapolation, which has more uncertainty, if it estimates a value outside of a range of given data.
- Any limitations or sources of uncertainty in the interpolated result
A limitation includes not accounting for any irregular changes between data and errors in original measurements.


Quantitative Short-TI Inversion Recovery (STIR Quant) was used as [measure]([https://pmc.ncbi.nlm.nih.gov/articles/PMC7190715/](https://pmc.ncbi.nlm.nih.gov/articles/PMC7190715/)) of inflammation and fluid buildup at the mid-point of the tibialis anterior. Similarly to fat fraction percentage, SITR Quant is a dimensionless quantitative anatomical measure that can demonstrate the beginnings of muscular dystrophy in FSHD patients. In addition to STIR Quant, dorsiflexion strength (measured in Newtons) was used in the following analysis as a functional measure of the patients' FSHD. Finally, the Clinical Severity Score on a scale of 0-10 was used as a clinical measure to compare with STIR Quant and dorsiflexion strength. All three measures were interpolated to the lowest period of time from baseline at T1 = 11.15 months and T2 =  22.13 months from the baseline measurement. If a patient did not have a T2 data point, they were removed from the T2 analysis. The following code imports and pulls the patient data from the CSV file, calculates the interpolations, and provides an example plot of the interpolation of four patients.

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator
from scipy import stats

CSV_PATH = "full analysis codes/Module2_FSHD_Dataset.csv"
PAUSE_AFTER_CHECK = True      # show the interpolation check plot before continuing
N_EXAMPLE_PATIENTS = 4        # number of patients drawn in the interpolation check
OUTLIER_SD = 2                # outlier cutoff in standard deviations

# load data ----------------------------------------------------------------
if not os.path.exists(CSV_PATH):
    folder = os.path.dirname(os.path.abspath(__file__)) if "__file__" in globals() else os.getcwd()
    candidate = os.path.join(folder, os.path.basename(CSV_PATH))
    if os.path.exists(candidate):
        CSV_PATH = candidate

df = pd.read_csv(CSV_PATH)

# Check for required numeric data in each patient in CSV file.
required = ["subject_id", "lag_months_y1", "lag_months_y2"]
missing = [c for c in required if c not in df.columns]
if missing:
    raise ValueError(f"Missing required columns: {missing}")
if df["subject_id"].duplicated().any():
    raise ValueError("subject_id must be unique; duplicate subject IDs were found.")

for c in ["lag_months_y1", "lag_months_y2"]:
    df[c] = pd.to_numeric(df[c], errors="coerce")

# Determine lowest lag times for interpolation of T1 and T2
y1_lags = df.loc[df["lag_months_y1"] > 0, "lag_months_y1"]
y2_lags = df.loc[df["lag_months_y2"] > 0, "lag_months_y2"]
if y1_lags.empty or y2_lags.empty:
    raise ValueError("Positive Year-1 and Year-2 visit lags are required.")
T1, T2 = float(y1_lags.min()), float(y2_lags.min())
if T2 <= T1:
    raise ValueError(f"T2 ({T2:g}) must be later than T1 ({T1:g}).")

TP_ALL = [("y1", T1), ("y2", T2)]
TP_STIR = [("y1", T1)]

# Display column names for csv files
TARGETS = [
    ("Dorsiflexion R (QMT)", "dorsiflexion_qmt_R_baseline", "dorsiflexion_qmt_R_y1", "dorsiflexion_qmt_R_y2", "dors_r", TP_ALL),
    ("Dorsiflexion L (QMT)", "dorsiflexion_qmt_L_baseline", "dorsiflexion_qmt_L_y1", "dorsiflexion_qmt_L_y2", "dors_l", TP_ALL),
    ("Clinical severity score", "clinical_severity_score_baseline", "clinical_severity_score_y1", "clinical_severity_score_y2", "severity", TP_ALL),
    ("STIR quant R", "stir_quant_baseline_R", "stir_quant_y1_R", "stir_quant_y2_R", "stir_r", TP_STIR),
    ("STIR quant L", "stir_quant_baseline_L", "stir_quant_y1_L", "stir_quant_y2_L", "stir_l", TP_STIR),
]

# Helper Methods ----------------------------------------------------------------
def number(row, col):
    """Return row value as float."""
    if col not in row.index:
        return np.nan
    v = pd.to_numeric(pd.Series([row[col]]), errors="coerce").iloc[0]
    return float(v) if pd.notna(v) and np.isfinite(v) else np.nan

def observations(row, base_col, y1_col, y2_col):
    """List patient (time, value, label) points: baseline, Y1, Y2."""
    raw = [
        (0.0, number(row, base_col), "baseline"),
        (number(row, "lag_months_y1"), number(row, y1_col), "Y1"),
        (number(row, "lag_months_y2"), number(row, y2_col), "Y2"),
    ]
    raw = [(float(t), float(v), label) for t, v, label in raw
           if pd.notna(t) and pd.notna(v) and np.isfinite(t) and np.isfinite(v) and t >= 0]
    # Average data points that share the same time (right and left leg)
    grouped = {}
    for t, v, label in raw:
        grouped.setdefault(t, []).append((v, label))
    return [(t, float(np.mean([v for v, _ in grouped[t]])),
             "+".join(label for _, label in grouped[t]))
            for t in sorted(grouped)]

def estimate(points, target):
    """Linear interpolation between two time points"""
    # Use observed value if one exists at the target time.
    for t, v, _ in points:
        if np.isclose(t, target, atol=1e-8):
            return v, "observed_exact", t, t
    if len(points) < 2:
        return np.nan, "insufficient_observations", np.nan, np.nan
    times = np.array([p[0] for p in points], dtype=float)
    values = np.array([p[1] for p in points], dtype=float)
    if target < times[0] or target > times[-1]:
        return np.nan, "outside_observed_range_no_extrapolation", np.nan, np.nan
    right = int(np.searchsorted(times, target, side="right"))
    left = right - 1
    t0, t1 = times[left], times[right]
    if t1 <= t0:
        return np.nan, "invalid_nonpositive_time_interval", t0, t1
    val = values[left] + (values[right] - values[left]) * ((target - t0) / (t1 - t0))
    return float(val), "linear_interpolation", float(t0), float(t1)

# Interpolation to T1 & T2 ----------------------------------------------------------------
standard = pd.DataFrame({"subject_id": df["subject_id"]})
audit_rows = []   # log of how every estimate was calculated
for display, base_col, y1_col, y2_col, key, tps in TARGETS:
    standard[f"{key}_baseline"] = np.nan
    for tp, _ in tps:
        standard[f"{key}_{tp}"] = np.nan
    for idx, row in df.iterrows():
        pts = observations(row, base_col, y1_col, y2_col)
        standard.loc[idx, f"{key}_baseline"] = number(row, base_col)
        for tp, target in tps:
            val, method, source_start, source_end = estimate(pts, target)
            standard.loc[idx, f"{key}_{tp}"] = val
            audit_rows.append({
                "subject_id": row["subject_id"], "measure": display,
                "target_timepoint": tp.upper(), "target_months": target,
                "estimated_value": val, "method": method,
                "source_start_months": source_start, "source_end_months": source_end,
                "observed_y1_lag_months": number(row, "lag_months_y1"),
                "observed_y2_lag_months": number(row, "lag_months_y2"),
                "valid_observation_count": len(pts),
            })

# pd.DataFrame(audit_rows).to_csv("interpolation_check.csv", index=False)
# standard.to_csv("interpolated_common_timepoints.csv", index=False)

# Interpolation Check Plot ----------------------------------------------------------------
# 4 example patients
# star = interpolated T1, diamond = interpolated T2
fig0, ax0 = plt.subplots(figsize=(8, 5.5))
shown = 0
for idx, row in df.iterrows():
    pts = observations(row, "dorsiflexion_qmt_R_baseline",
                       "dorsiflexion_qmt_R_y1", "dorsiflexion_qmt_R_y2")
    if len(pts) < 2:
        continue
    line, = ax0.plot([p[0] for p in pts], [p[1] for p in pts], marker="o",
                     linewidth=1.2, label=f"Subject {row['subject_id']}")
    v1, v2 = standard.loc[idx, "dors_r_y1"], standard.loc[idx, "dors_r_y2"]
    if pd.notna(v1):
        ax0.scatter(T1, v1, marker="*", s=180, color=line.get_color(), edgecolor="black", zorder=4)
    if pd.notna(v2):
        ax0.scatter(T2, v2, marker="D", s=70, color=line.get_color(), edgecolor="black", zorder=4)
    shown += 1
    if shown >= N_EXAMPLE_PATIENTS:
        break
ax0.axvline(T1, color="gray", linestyle="--", label=f"T1={T1:g} months")
ax0.axvline(T2, color="gray", linestyle=":", label=f"T2={T2:g} months")
ax0.set_xlabel("Months since baseline")
ax0.set_ylabel("Right Dorsiflexion Strength (N)")
ax0.set_title(f"Interpolation Example ({shown} patients; star = T1, diamond = T2)")
ax0.legend(fontsize=8)
fig0.tight_layout()
# fig0.savefig("interpolation_check.png", dpi=150)
if PAUSE_AFTER_CHECK:
    plt.show()
```

After interpolation, an outlier analysis was performed on each measurement type and time point. Patients that did not have a baseline, T1, or T2 datapoint were not included in the sample for the corresponding measuremnt/change in time. For each sample, the mean and standard deviation were calculated. Then, the z-score of each patient was calculated. For this analysis, outliers were considered to be patients that had a z-score greater than two standard deviations away from the mean. Outliers were then removed from the sample for the final linear regression and mean % change analysis.

```python
# Outlier Analysis on Interpolated Data Points ----------------------------------------------------------------
NAMES = {"dors": "Dorsiflexion", "stir": "STIR quant", "severity": "Clinical severity"}
COL_NAMES = {"dors_r": "Dorsiflexion R", "dors_l": "Dorsiflexion L", "severity": "Clinical severity",
             "stir_r": "STIR quant R", "stir_l": "STIR quant L"}
TP_NAMES = {"baseline": "Baseline", "y1": "T1", "y2": "T2"}
test_cols = [c for c in standard.columns if c != "subject_id"]

outlier_masks = {}
for col in test_cols:
    vals = standard[col]
    sd = vals.std(ddof=1)
    if vals.notna().sum() > 2 and sd > 0:
        outlier_masks[col] = ((vals - vals.mean()) / sd).abs() > OUTLIER_SD
    else:
        outlier_masks[col] = pd.Series(False, index=standard.index)

print(f"\nOutlier test on interpolated values (|z| > {OUTLIER_SD}), patients removed per measure/time point:")
outlier_rows = []
for col in test_cols:
    mask = outlier_masks[col]
    mkey, tp = col.rsplit("_", 1) if col.rsplit("_", 1)[-1] in ("y1", "y2") else (col.rsplit("_", 1)[0], "baseline")
    n_before = int(standard[col].notna().sum())
    outlier_rows.append({"measure": COL_NAMES[mkey], "time_point": TP_NAMES[tp],
                         "n_before": n_before, "n_removed": int(mask.sum()),
                         "n_remaining": n_before - int(mask.sum()),
                         "removed_subject_ids": ", ".join(map(str, df.loc[mask, "subject_id"]))})
any_outlier = pd.concat(outlier_masks.values(), axis=1).any(axis=1)
clean = standard.copy()
for col in test_cols:
    clean.loc[outlier_masks[col], col] = np.nan
outlier_table = pd.DataFrame(outlier_rows)
outlier_table.to_csv("outlier_summary.csv", index=False)
clean.to_csv("interpolated_common_timepoints_outliers_removed.csv", index=False)
print(outlier_table.to_string(index=False))
print(f"Total value removals: {int(outlier_table['n_removed'].sum())}; "
      f"unique patients with at least one removal: {int(any_outlier.sum())}")
```

Next, the percent change of each measurement type was calculated for baseline to T1, baseline to T2, and T1 to T2. These calculated values are then used for statistical analysis below.

```python
# Percent Changes ----------------------------------------------------------------
COMPARISONS = [
    ("Baseline-T1", "baseline", "y1", T1, "b_t1"),
    ("Baseline-T2", "baseline", "y2", T2, "b_t2"),
    ("T1-T2", "y1", "y2", T2 - T1, "t1_t2"),
]
if any(c[3] <= 0 for c in COMPARISONS):
    raise ValueError("All comparison time differences must be positive.")

def percent_change(earlier, later):
    """Percent change relative to |earlier|; a zero baseline gives NaN (avoids divide-by-zero)."""
    return (later - earlier) / earlier.abs().replace(0, np.nan) * 100

def limb_average(data, right, left, early, late):
    """Average right and left limb percent changes into one value per patient
    (uses the single available limb if the other is missing)."""
    r = percent_change(data[f"{right}_{early}"], data[f"{right}_{late}"])
    l = percent_change(data[f"{left}_{early}"], data[f"{left}_{late}"])
    return pd.concat([r, l], axis=1).mean(axis=1, skipna=True)

# One row per patient. Columns are named <measure>_<comparison>, e.g. dors_b_t1.
pct = pd.DataFrame({"subject_id": df["subject_id"]})
for _, early, late, _, ckey in COMPARISONS:
    pct[f"dors_{ckey}"] = limb_average(clean, "dors_r", "dors_l", early, late)
    pct[f"severity_{ckey}"] = percent_change(clean[f"severity_{early}"], clean[f"severity_{late}"])
pct["stir_b_t1"] = limb_average(clean, "stir_r", "stir_l", "baseline", "y1")   # STIR: Baseline-T1 only
pct = pct.replace([np.inf, -np.inf], np.nan)
pct.to_csv("percent_differences.csv", index=False)
```

The first statistical analyses computed were linear regressions with corresponding scatter plots. For each time difference, the percent change of dorsiflexion strength or STIR Quant was plotted against the Clinical Severity Score. A linear regression for STIR Quant was only computed for the percent change from baseline to T1, since there were no T2 values collected. An R-squared value was calculated for each regression, demonstrating the strength of the correlation between the two measures. The results of the statistical analysis are printed for each measurement type and time point difference, where n is the sample size, pearson_r is the Pearson correlation coefficient, and R2 is the R-squared value of regression. The R-squared value, in addition to the linear regression equation, is also displayed on each plot.

```python
# Linear Regression and Scatter Plots ----------------------------------------------------------------
fig1, axes = plt.subplots(2, 3, figsize=(17, 10))
panels = [(axes[0, i], "dors", "Dorsiflexion", c[0], c[4]) for i, c in enumerate(COMPARISONS)]
panels.append((axes[1, 0], "stir", "STIR quant", "Baseline-T1", "b_t1"))
axes[1, 1].set_axis_off()
axes[1, 2].set_axis_off()

reg_rows = []
for ax, mkey, mname, label, ckey in panels:
    pair = pct[[f"{mkey}_{ckey}", f"severity_{ckey}"]].dropna()
    pair.columns = ["x", "y"]
    row = {"measure": mname, "comparison": label, "n": len(pair),
           "slope": np.nan, "pearson_r": np.nan, "R2": np.nan}
    ax.axhline(0, color="gray", linewidth=0.6)
    ax.axvline(0, color="gray", linewidth=0.6)
    ax.set_xlabel(f"{mname} percent change ({label})")
    ax.set_ylabel(f"Clinical severity percent change ({label})")

    # Need enough varied points to fit a line.
    if len(pair) < 3 or pair["x"].nunique() < 2 or pair["y"].nunique() < 2:
        ax.set_title(f"{mname}: {label} (n={len(pair)})")
        ax.text(0.5, 0.5, "Insufficient paired data for regression",
                ha="center", va="center", transform=ax.transAxes)
        reg_rows.append(row)
        continue

    ax.scatter(pair["x"], pair["y"], edgecolor="black", linewidth=0.4, alpha=0.75)
    slope, intercept = np.polyfit(pair["x"], pair["y"], 1)       # least-squares line
    predicted = slope * pair["x"] + intercept
    ss_res = ((pair["y"] - predicted) ** 2).sum()
    ss_tot = ((pair["y"] - pair["y"].mean()) ** 2).sum()
    r2 = 1 - ss_res / ss_tot if ss_tot else np.nan
    xs = np.linspace(pair["x"].min(), pair["x"].max(), 100)
    ax.plot(xs, slope * xs + intercept, color="black",
            label=f"y={slope:.3f}x {intercept:+.2f}\nR\u00b2={r2:.3f}")
    ax.legend(fontsize=8)
    ax.set_title(f"{mname}: {label} (n={len(pair)})")
    row.update(slope=slope, pearson_r=np.corrcoef(pair["x"], pair["y"])[0, 1], R2=r2)
    reg_rows.append(row)

fig1.suptitle("Clinical severity percent change vs. dorsiflexion / STIR quant")
fig1.tight_layout()
fig1.savefig("percent_difference_plots.png", dpi=150)

reg = pd.DataFrame(reg_rows)
reg.to_csv("regression_results.csv", index=False)
print("\nRegression of severity % change on each measure's % change:")
print(reg.round(4).to_string(index=False))
```

To analytically describe the stastical signifance of the bar plots we conducts a one-sample t-test. For every plot the p-values were greater than 0.05, therefore there is so stastical signifance between each plot

```python
# ---------------------------------------------------------------- mean +/- SD plots
# One separate plot (panel) per measure. Each bar = mean, error bar = SD,
# dots = individual patients. STIR only has Baseline-T1.
# A one-sample t-test (H0: mean percent change = 0) is run for every bar.
fig2, axes2 = plt.subplots(1, 3, figsize=(17, 6.5))
rng = np.random.default_rng(3)
summary_rows = []
for ax, mkey in zip(axes2, ["dors", "stir", "severity"]):
    mname = NAMES[mkey]
    comps = [c for c in COMPARISONS if mkey != "stir" or c[4] == "b_t1"]
    stat_trans = ax.get_xaxis_transform()        # x in data units, y in axes fraction
    ax.axhline(0, color="gray", linewidth=0.6)
    lo_all, hi_all = np.inf, -np.inf
    for i, (label, early, late, elapsed, ckey) in enumerate(comps):
        values = pct[f"{mkey}_{ckey}"].dropna()
        mean = values.mean() if len(values) else np.nan
        sd = values.std(ddof=1) if len(values) > 1 else np.nan

        # One-sample t-test against 0 (needs at least 2 values and some variance).
        if len(values) > 1 and values.nunique() > 1:
            t_stat, p_val = stats.ttest_1samp(values, 0.0)
        else:
            t_stat, p_val = np.nan, np.nan

        summary_rows.append({"measure": mname, "comparison": label, "elapsed_months": elapsed,
                             "n": len(values), "mean_percent_change": mean, "sd_percent_change": sd,
                             "t_stat": t_stat, "p_value": p_val})
        if values.empty:
            continue
        sd0 = 0.0 if np.isnan(sd) else sd
        ax.bar(i, mean, width=0.6, alpha=0.5, edgecolor="black")
        ax.errorbar(i, mean, yerr=sd0, fmt="none", ecolor="black", capsize=6)
        ax.scatter(i + rng.uniform(-0.15, 0.15, len(values)), values,
                   s=18, color="black", alpha=0.5, zorder=3)
        # n / mean / SD / p printed above the plot area so they never overlap the data.
        p_txt = "p=n/a" if np.isnan(p_val) else (f"p={p_val:.3f}" if p_val >= 0.001 else "p<0.001")
        ax.text(i, 1.01, f"n={len(values)}\nmean={mean:.1f}\nSD={sd0:.1f}\n{p_txt}",
                transform=stat_trans, ha="center", va="bottom", fontsize=9)
        lo_all = min(lo_all, values.min(), mean - sd0)
        hi_all = max(hi_all, values.max(), mean + sd0)

    # Zoom the y-axis to this measure's data (points and SD bars) so extremes sit near top/bottom.
    if np.isfinite(lo_all):
        pad = 0.04 * (hi_all - lo_all) if hi_all > lo_all else 1.0
        ax.set_ylim(lo_all - pad, hi_all + pad)
    ax.yaxis.set_major_locator(MaxNLocator(nbins=15))   # more y-axis tick marks
    ax.grid(axis="y", alpha=0.3)
    ax.set_xticks(range(len(comps)))
    ax.set_xticklabels([f"{c[0]}\n({c[3]:g} mo)" for c in comps], fontsize=10)
    ax.set_xlim(-0.6, len(comps) - 0.4)
    ax.set_ylabel("Percent change (%)")
    ax.set_title(f"{mname} percent change", pad=70)   # extra room for the p-value line
fig2.suptitle("Mean \u00b1 SD of patient-level percent changes (limbs averaged); "
              "p = one-sample t-test vs 0")
fig2.tight_layout()
fig2.savefig("mean_sd_plots.png", dpi=150)

summary = pd.DataFrame(summary_rows)
summary.to_csv("summary_stats.csv", index=False)
print("\nSummary statistics (one-sample t-test of percent change vs 0):")
print(summary.round({"elapsed_months": 2, "mean_percent_change": 2, "sd_percent_change": 2,
                     "t_stat": 3, "p_value": 4}).to_string(index=False))
plt.show()
```

## MRI Data Analysis Methdology:
Many MRI image analyzes of FSHD involved calculating fat fraction ratios. Although the above interpolation analysis did not utilize fat fraction percentage, the following code is a computational method to analyze an MRI image for the fat fraction in the tibialis anterior muscle.

First, individual rows for water, fat, and mask were loaded in a variables and printed:

```python
# IMPORTING PACKAGES
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

DATA_DIR = Path("/Users/charlottelim/Library/CloudStorage/OneDrive-UniversityofVirginia/fall ‘26/bme 2315/module 2/bme2315module2/FSHD_MRI_data_char/FSHD_TA_MRI_data_0020_TP1_TP2") # must match path to personal computer
manifest = pd.read_csv(DATA_DIR / "slice_info.csv")
manifest

row = manifest[manifest["timepoint"] == "TP1"].iloc[3]   # representative mid-muscle slice

water = np.load(DATA_DIR / row["water_file"]) # water signal intensity per pixel
fat = np.load(DATA_DIR / row["fat_file"]) # fat signal intensity at each pixel
mask = np.load(DATA_DIR / row["mask_file"]).astype(bool) # finds muscle pixels

print("Water image shape:", water.shape)
print("Fat image shape:", fat.shape)
print("Mask shape:", mask.shape)
print("Number of muscle pixels:", mask.sum())
```

The shape of the muscle was printed used “.shape”  method after the variable by and  then using the sum method on the number of pixels or data points to get the sum of the pixels of the TA muscle. The fat fraction was calculated from intensities of each pixel which indicated fat and water signals within the TA muscle shape. For the whole image region, all pixels were includes. For TA muscle the mask was applied.

```python
denominator = water + fat

valid_pixels = denominator > 0

fat_fraction = np.divide( # fat % at each pixel
    fat,
    denominator,
    out=np.zeros_like(fat, dtype=float),
    where=valid_pixels
) * 100
fat_fraction_full = fat_fraction.copy()

fat_fraction_masked = fat_fraction.copy()
fat_fraction_masked[~mask] = np.nan

whole_image_values = fat_fraction_full[valid_pixels]
muscle_values = fat_fraction_full[mask & valid_pixels]

muscle_fat_percentage = np.mean(muscle_values)

print("Number of Whole Image Pixels: " + str(whole_image_values.size))
print("Number of Muscle Pixels: " + str(muscle_values.size ))
print("Average Muscle Fat Percentage: "+ str(muscle_fat_percentage) + "%")
```

As shown in the code above, fat fraction distrubutition was summarized by mean, minimum and maximum fat-fraction percentages for both the whole image and TA muscle region. All zero vallues were excluded. The final average fat percentage for the entire muscle is as shown above.
