* Symptoms: Facial weakness is often the earliest sign of FSHD, making it hard to use a straw or smile. Other hallmark symptoms include shoulder weakness, causing scapular winging, and lower leg weakness, causing a foot drop which increases the risk of tripping and difficulty walking. Additionally, chronic pain is common in FSHD patients due to strain related to posture and weakness. Source: https://www.mda.org/disease/facioscapulohumeral-muscular-dystrophy/signs-and-symptoms
* Diagnosis: Diagnosis of FSHD is usually suspected based on a pattern of muscle weakness, typically in the face shoulder blades and upper arms, often with asymmetric involvement. The most reliable way to confirm a diagnosis is through genetic testing. A contraction of the D4Z4 region on chromosome 4 identifies this disease. In some cases where a diagnosis is unclear, blood tests that measure creatine kinase, EMGs, or muscle biopsies can also be used. ## MRI Data Analysis Methdology:
Many MRI image analyzes of FSHD involved calculating fat fraction ratios. The following is an computational method to anaylze an MRI image for the fat fraction in the tibialis anterior muscle.

First, individual rows for water, fat and mask were loaded in a variables and printed: 

row = manifest[manifest["timepoint"] == "TP1"].iloc[3]   # representative mid-muscle slice

water = np.load(DATA_DIR / row["water_file"])
fat = np.load(DATA_DIR / row["fat_file"])
mask = np.load(DATA_DIR / row["mask_file"]).astype(bool)

print("Water image shape:", water.shape)
print("Fat image shape:", fat.shape)
print("Mask shape:", mask.shape)
print("Number of muscle pixels:", mask.sum())

The shape of the muscle was printed used “.shape”  method after the variable by and  then using the sum method on the number of pixels or data points to get the sum of the pixels of the TA muscle. The fat fraction was calculated from intensities of each pixel which indicated fat and water signals within the TA muscle shape. For the whole image region, all pixels were includes. For TA muscle the mask was applied.

denominator = water + fat

valid_pixels = denominator > 0

fat_fraction = np.divide(
    fat,
    denominator,
    out=np.zeros_like(fat, dtype=float),
    where=valid_pixels
) * 100
fat_fraction_full = fat_fraction.copy()

fat_fraction_masked = fat_fraction.copy()
fat_fraction_masked[~mask] = np.nan

whole_image_values = fat_fraction_full[valid_pixels]
muscle_values = fat_fraction_masked[~np.isnan(fat_fraction_masked)]

As shown in the code above, fat fraction distrubutition was summarized by mean, minimum and maximum fat-fraction percentages for both the whole image and TA muscle region. All zero vallues were excluded.
